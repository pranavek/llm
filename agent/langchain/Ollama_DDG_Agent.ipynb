```python
# Import relevant functionality
# https://github.com/langchain-ai/langchain/issues/21479
# https://python.langchain.com/docs/tutorials/agents/
from langchain_openai import ChatOpenAI
from langchain_community.tools.ddg_search.tool import DuckDuckGoSearchResults
from langchain_community.tools.file_management.list_dir import ListDirectoryTool
from langchain_core.messages import HumanMessage
from langgraph.checkpoint.memory import MemorySaver
from langgraph.prebuilt import create_react_agent

# Create the agent
memory = MemorySaver()
llm = ChatOpenAI(
    api_key="ollama",
    model="qwen2.5",
    base_url="http://localhost:11434/v1",
)
search = DuckDuckGoSearchResults(max_results=2)
ls = ListDirectoryTool()

tools = [search, ls]
agent_executor = create_react_agent(llm, tools, checkpointer=memory)

# Use the agent
config = {"configurable": {"thread_id": "abc123"}}
for step in agent_executor.stream(
    {"messages": [HumanMessage(content="I'm living in birmingham, alabaama. whats the weather where I live?")]},
    config,
    stream_mode="values",
):
    step["messages"][-1].pretty_print()

for step in agent_executor.stream(
    {"messages": [HumanMessage(content="list files in /Users/pellathkizhakkayil/pranav/llm")]},
    config,
    stream_mode="values",
):
    step["messages"][-1].pretty_print()
```
